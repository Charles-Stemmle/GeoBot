```python
import pandas as pd
import matplotlib.pyplot as plt

image_data = pd.read_csv("../dataset.csv")
```

```python
country_counts = image_data["country"].value_counts()

top_10 = country_counts.head(10)

plt.bar(top_10.index, top_10.values)

plt.xticks(rotation=45, ha="right")

plt.show()
```

```python
#print countries with over 20 data entries 

countries_over_20 = country_counts[country_counts > 20]
print(countries_over_20)
```

```python
few_images = country_counts[country_counts < 20]

print(few_images)
print("Number of countries:", len(few_images))
```

```python
#investigate cut off rate for countries with fewer than n images

for minimum in [5, 10, 20, 50, 100]:
    countries = country_counts[country_counts >= minimum]
    print(minimum, len(countries))
```

```python
minimum_images = 50

countries_to_keep = country_counts[
    country_counts >= minimum_images
].index

print(countries_to_keep)
print("Countries kept:", len(countries_to_keep))
```

```python
filtered_data = image_data[image_data["country"].isin(countries_to_keep)]
print(filtered_data.shape)
print(filtered_data["country"].value_counts())
```

```python
filtered_data.to_csv("filtered_dataset.csv", index=False)
```
